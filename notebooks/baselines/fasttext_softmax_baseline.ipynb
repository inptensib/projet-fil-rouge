```python
# --- Limiter BLAS à 1 thread par worker (évite N_JOBS × BLAS_THREADS) ---
import os
os.environ["OMP_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"
os.environ["VECLIB_MAXIMUM_THREADS"] = "1"
os.environ["NUMEXPR_NUM_THREADS"] = "1"

# --- Configuration ---
import sys, time, warnings, tempfile, os
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import f1_score
from IPython.display import display

warnings.filterwarnings("ignore")
%matplotlib inline

from rakuten.common.data_loader import prepare_hierarchical_data_from_splits
from rakuten.common.metrics import (compute_all_metrics, compute_per_class_f1,
                             plot_confusion_matrix, plot_per_class_f1)
from rakuten.common.utils import set_seed

SEED = 42
N_JOBS = int(os.environ.get("N_JOBS", 26))  # Défaut 26, surchargeable via env
set_seed(SEED)

PROJECT_ROOT = Path.cwd().parent.parent
DATASET_DIR = Path("/home/infres/tandriam-25/fil_rouge/dataset")
SAVE_DIR = PROJECT_ROOT / "models" / "baselines" / "fasttext_softmax"
RESULTS_DIR = PROJECT_ROOT / "results" / "baselines" / "fasttext_softmax"
SAVE_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

sns.set_theme(style="whitegrid", font_scale=1.1)
plt.rcParams.update({"figure.dpi": 120, "figure.facecolor": "white"})

# --- Fix compatibilité NumPy 2.x / fasttext ---
import fasttext
def _patched_predict(self, text, k=1, threshold=0.0, on_unicode_error="strict"):
    def check(entry):
        if entry.find("\n") != -1:
            raise ValueError("predict processes one line at a time (remove '\\n')")
        entry += "\n"
        return entry

    if type(text) == list:
        text = [check(entry) for entry in text]
        all_labels, all_probs = self.f.multilinePredict(text, k, threshold, on_unicode_error)
        return all_labels, all_probs
    else:
        text = check(text)
        predictions = self.f.predict(text, k, threshold, on_unicode_error)
        if predictions:
            probs, labels = zip(*predictions)
        else:
            probs, labels = ([], ())
        return labels, np.asarray(probs)

fasttext.FastText._FastText.predict = _patched_predict

print(f"Configuration terminée. N_JOBS={N_JOBS}")
print(f"Modèles → {SAVE_DIR}")
print(f"Résultats → {RESULTS_DIR}")
```

## 2. Préparation des données FastText

FastText attend un fichier texte avec le format : `__label__<classe> <texte>`

Chaque ligne contient un label préfixé par `__label__` suivi du texte du produit.

```python
# --- Conversion au format FastText ---
def write_fasttext_file(texts, labels, encoder, filepath):
    """Écrit les données au format FastText."""
    label_names = encoder.inverse_transform(labels)
    with open(filepath, "w", encoding="utf-8") as f:
        for text, label in zip(texts, label_names):
            clean_label = str(label).replace(" ", "_").replace("\n", "_")
            clean_text = str(text).replace("\n", " ").strip()
            f.write(f"__label__{clean_label} {clean_text}\n")

tmpdir = tempfile.mkdtemp()

# Fichiers L1
train_l1_path = os.path.join(tmpdir, "train_l1.txt")
test_l1_path = os.path.join(tmpdir, "test_l1.txt")
write_fasttext_file(data["X_train"], data["y_l1_train"], data["l1_encoder"], train_l1_path)
write_fasttext_file(data["X_test"], data["y_l1_test"], data["l1_encoder"], test_l1_path)

# Fichiers L2
train_l2_path = os.path.join(tmpdir, "train_l2.txt")
test_l2_path = os.path.join(tmpdir, "test_l2.txt")
write_fasttext_file(data["X_train"], data["y_l2_train"], data["l2_encoder"], train_l2_path)
write_fasttext_file(data["X_test"], data["y_l2_test"], data["l2_encoder"], test_l2_path)

# Vérification
for name, path in [("train_l1", train_l1_path), ("test_l1", test_l1_path),
                    ("train_l2", train_l2_path), ("test_l2", test_l2_path)]:
    n_lines = sum(1 for _ in open(path))
    print(f"  {name}: {n_lines:,} lignes")
```

<a id="section-3"></a>
## 3. Classification L1 (157 catégories)

Entraînement avec `loss="softmax"` (identique au notebook FastText OVA).
Le modèle L1 existant est réutilisé s'il est disponible pour éviter un ré-entraînement inutile.

```python
# --- Entraînement FastText L1 ---
# Réutilise le modèle L1 OVA si disponible (même config : softmax, epoch=25, dim=100)
_l1_path = SAVE_DIR / "model_l1.bin"
_l1_path_original = PROJECT_ROOT / "models" / "baselines" / "fasttext" / "model_l1.bin"

if _l1_path.exists():
    model_l1 = fasttext.load_model(str(_l1_path))
    print(f"Modèle L1 chargé depuis {_l1_path}")
    train_time_l1 = 0
elif _l1_path_original.exists():
    model_l1 = fasttext.load_model(str(_l1_path_original))
    print(f"Modèle L1 chargé depuis {_l1_path_original} (modèle original réutilisé)")
    train_time_l1 = 0
else:
    print(f"Entraînement FastText L1 ({N_JOBS} threads)...")
    t0 = time.time()
    model_l1 = fasttext.train_supervised(
        input=train_l1_path,
        lr=0.5, epoch=25, wordNgrams=2, dim=100,
        loss="softmax", thread=N_JOBS, verbose=0,
    )
    train_time_l1 = time.time() - t0
    model_l1.save_model(str(_l1_path))
    print(f"Entraînement L1 terminé en {train_time_l1:.0f}s — sauvegardé dans {_l1_path}")

# --- Prédiction batch ---
def fasttext_predict_batch(model, texts, encoder, label=""):
    """Prédit avec FastText en batch et convertit en indices encodés.

    Normalise en NFC les deux côtés du mapping pour corriger les décalages
    d'encodage Unicode (Polars vs Python str) entre l'entraînement et l'éval.
    """
    import unicodedata
    class_to_idx_nfc = {
        unicodedata.normalize('NFC', c): i
        for i, c in enumerate(encoder.classes_)
    }
    ft_to_idx = {}
    unmapped = []
    for ft_label in model.labels:
        name = unicodedata.normalize('NFC',
                    ft_label.replace("__label__", "").replace("_", " "))
        if name in class_to_idx_nfc:
            ft_to_idx[ft_label] = class_to_idx_nfc[name]
        else:
            unmapped.append(ft_label)
            ft_to_idx[ft_label] = 0
    if unmapped:
        print(f"  ⚠ {len(unmapped)} labels non mappés (fallback 0) : {unmapped[:3]}")
    else:
        print(f"  ✓ Tous les {len(ft_to_idx)} labels mappés")

    cleaned = [str(t).replace("\n", " ").strip() for t in texts]
    print(f"  {label} prédiction batch sur {len(cleaned):,} textes...")
    labels_batch, _ = model.predict(cleaned)
    preds = [ft_to_idx.get(lbs[0], 0) for lbs in labels_batch]
    return np.array(preds)

print("Prédiction L1 sur le test set...")
preds_l1_test = fasttext_predict_batch(model_l1, data["X_test"], data["l1_encoder"], label="L1")
f1_l1 = f1_score(data["y_l1_test"], preds_l1_test, average="macro")
print(f"FastText L1 — Test F1 macro : {f1_l1:.4f}")
```

<a id="section-4"></a>
## 4. Classification L2 (6000+ sous-catégories) — Softmax

Contrairement au notebook FastText OVA, on utilise ici `loss="softmax"` pour L2.
Le softmax est plus adapté à ce problème **single-label** : les 6 000+ classes se font
concurrence directement, comme dans HYDRA et TF-IDF, au lieu d'être traitées
comme des classificateurs binaires indépendants.

```python
# --- Entraînement FastText L2 (softmax) ---
_l2_path = SAVE_DIR / "model_l2.bin"
L2_THREADS = min(N_JOBS, 10)

if _l2_path.exists():
    model_l2 = fasttext.load_model(str(_l2_path))
    print(f"Modèle L2 chargé depuis {_l2_path}")
    train_time_l2 = 0
else:
    print(f"Entraînement FastText L2 softmax (6000+ classes, {L2_THREADS} threads)...")
    t0 = time.time()
    model_l2 = fasttext.train_supervised(
        input=train_l2_path,
        lr=0.5, epoch=25, wordNgrams=2, dim=100,
        loss="softmax", thread=L2_THREADS, verbose=0,
    )
    train_time_l2 = time.time() - t0
    model_l2.save_model(str(_l2_path))
    print(f"Entraînement L2 terminé en {train_time_l2:.0f}s — sauvegardé dans {_l2_path}")

preds_l2_test = fasttext_predict_batch(model_l2, data["X_test"], data["l2_encoder"], label="L2")
f1_l2 = f1_score(data["y_l2_test"], preds_l2_test, average="macro")
print(f"FastText L2 Softmax — Test F1 macro : {f1_l2:.4f}")
```

<a id="section-5"></a>
## 5. Métriques hiérarchiques

FastText entraîne L1 et L2 indépendamment — il n'y a aucune contrainte hiérarchique.

```python
# --- Métriques hiérarchiques ---
all_metrics = compute_all_metrics(
    labels_l1=data["y_l1_test"], preds_l1=preds_l1_test,
    labels_l2=data["y_l2_test"], preds_l2=preds_l2_test,
    l2_to_l1_mapping=data["l2_to_l1_mapping"],
)

print("=== Métriques FastText Softmax (Test Set) ===")
for k, v in all_metrics.items():
    if isinstance(v, float):
        print(f"  {k:30s} : {v:.4f}")

# Tableau récapitulatif
summary = pd.DataFrame([{
    "Modèle": "FastText Softmax",
    "F1 L1 macro": all_metrics.get("f1_l1_macro", f1_l1),
    "F1 L2 macro": all_metrics.get("f1_l2_macro", f1_l2),
    "HCR": all_metrics.get("hcr", 0),
    "Temps total": f"{train_time_l1 + train_time_l2:.0f}s",
}])
display(summary.style.format({
    "F1 L1 macro": "{:.4f}", "F1 L2 macro": "{:.4f}", "HCR": "{:.4f}",
}).set_caption("Résultats FastText Softmax (Test Set)"))
```

## 7. Conclusion

### Softmax vs OVA pour FastText L2
- **OVA** : 6 126 classificateurs binaires indépendants (sigmoid), conçu pour le multi-label.
  Inadapté ici car Rakuten L2 est single-label — résultat : F1 L2 ≈ 10%.
- **Softmax** : compétition directe entre toutes les classes, comme HYDRA et TF-IDF.
  Résultat attendu : F1 L2 entre 30% et 40%.

### Forces de FastText
- **Ultra-rapide** : entraînement en quelques minutes même sur 5.5M de textes
- **Sous-mots** : capture les morphèmes (préfixes, suffixes) et résiste aux fautes d'orthographe
- **Modèle intégré** : embedding + classifieur entraînés conjointement (end-to-end)

### Limites
- **Pas contextuel** : un mot a toujours le même vecteur quel que soit le contexte
- **Pas hiérarchique** : L1 et L2 sont entraînés indépendamment
- **Classifieur linéaire** : ne capture pas les interactions complexes entre features
